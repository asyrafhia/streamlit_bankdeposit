```python
import pandas as pd
from category_encoders.one_hot import OneHotEncoder
import joblib
```

```python
columns = [
    'age',
    'job',
    'marital',
    'education',
    'default',
    'balance',
    'housing',
    'loan',
    'contact',
    'month',
    'duration',
    'campaign',
    'previous',
]

# values = [
#     [35, 'admin.', 'married', 'secondary', 'no', 214,
#      'yes', 'yes', 'unknown', 'jun', 247, 1, 0]
# ]

values = [
    [50, 'management', 'married', 'tertiary', 'yes',
     5000, 'yes', 'yes', 'cellular', 'feb',
     2000, 2, 2]
]

X_test = pd.DataFrame(values, columns=columns)
X_test
```

```python
# import one hot encoder
ohe = joblib.load(MODEL_PATH + 'ohe.pkl')
ohe
```

```python
X_test_enc = ohe.transform(X_test)
X_test_enc
```

```python
# import cbc_optuna
cbc_optuna = joblib.load(MODEL_PATH + 'cbc_optuna.pkl')
cbc_optuna
```

```python
best_thres_accuracy = 0.4963388427754832
best_thres_accuracy
```

```python
model_proba = cbc_optuna.predict_proba(X_test_enc)

# Ambil probabilitas kelas 1
prob = model_proba[0][1]

# Prediksi berdasarkan threshold
model_pred = 'The client is predicted not to subscribe a term deposit.' if prob < best_thres_accuracy else 'The client is predicted to subscribe a term deposit.'

percentage = model_proba[0][1] * 100

print(f"Probability Percentage = {percentage:.2f}%")
print(model_pred)
```
